```python

# EXPERIMENT 7: COVID-19 TWEET SENTIMENT ANALYSIS

# STEP 1: IMPORT LIBRARIES

!pip -q install textblob wordcloud

import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns

from textblob import TextBlob
from wordcloud import WordCloud, STOPWORDS
from google.colab import files

pd.set_option("display.max_colwidth", None)

print("Libraries imported successfully!")

# STEP 2: UPLOAD AND LOAD DATASET

print("\nSTEP 2: UPLOAD COVID-19 TWEET DATASET")

uploaded = files.upload()

csv_files = [name for name in uploaded.keys() if name.lower().endswith(".csv")]

if not csv_files:
    raise ValueError("Please upload a CSV file.")

file_name = csv_files[0]
tweet_df = pd.read_csv(file_name)

# Identify required columns
tweet_column = next(
    (col for col in tweet_df.columns if col.lower() == "originaltweet"),
    None
)

sentiment_column = next(
    (col for col in tweet_df.columns if col.lower() == "sentiment"),
    None
)

if tweet_column is None:
    raise ValueError("OriginalTweet column not found in uploaded CSV.")

if sentiment_column is None:
    raise ValueError("Sentiment column not found in uploaded CSV.")

print("Dataset loaded successfully!")
print("File name:", file_name)
print("Total records:", len(tweet_df))
print("Columns:", tweet_df.columns.tolist())

# STEP 3: DISPLAY DATASET

print("\nSTEP 3 OUTPUT: FIRST 10 RECORDS")

display(tweet_df.head(10))

# STEP 4: CLEAN TWEETS

print("\nSTEP 4: CLEANING TWEETS")

def clean_tweets(tweet):
    tweet = str(tweet).lower()
    tweet = re.sub(r"@[A-Za-z0-9_]+", " ", tweet)
    tweet = re.sub(r"https?://\S+|www\.\S+", " ", tweet)
    tweet = re.sub(r"rt\s+", " ", tweet)
    tweet = re.sub(r"[^a-zA-Z\s]", " ", tweet)
    tweet = " ".join(tweet.split())
    return tweet

tweet_df["clean_tweets"] = tweet_df[tweet_column].fillna("").apply(clean_tweets)

print("\nSTEP 4 OUTPUT: CLEANED TWEETS")

display(tweet_df[[tweet_column, "clean_tweets"]].head(10))

# STEP 5: PERFORM SENTIMENT ANALYSIS USING TEXTBLOB

print("\nSTEP 5: TEXTBLOB SENTIMENT ANALYSIS")

def analyze_tweets(tweet):
    analysis = TextBlob(str(tweet))

    if analysis.sentiment.polarity > 0:
        return "Positive"
    elif analysis.sentiment.polarity < 0:
        return "Negative"
    else:
        return "Neutral"

tweet_df["TextBlob_Sentiment"] = tweet_df["clean_tweets"].apply(analyze_tweets)

print("\nSTEP 5 OUTPUT: SENTIMENT ANALYSIS")

display(tweet_df[["clean_tweets", "TextBlob_Sentiment"]].head(10))

# STEP 6: DISPLAY ORIGINAL SENTIMENT AND TEXTBLOB SENTIMENT

print("\nSTEP 6 OUTPUT: DATASET SENTIMENT VS TEXTBLOB SENTIMENT")

display(
    tweet_df[
        [tweet_column, sentiment_column, "TextBlob_Sentiment"]
    ].head(10)
)

# STEP 7: DISPLAY SAMPLE TWEET ANALYSIS

print("\nSTEP 7 OUTPUT: SAMPLE TWEET ANALYSIS")

n = min(5, len(tweet_df) - 1)

print("\nOriginal Tweet:")
print(tweet_df[tweet_column].iloc[n])

print("\nCleaned Tweet:")
print(tweet_df["clean_tweets"].iloc[n])

print("\nTextBlob Sentiment:")
print(tweet_df["TextBlob_Sentiment"].iloc[n])

# STEP 8: CALCULATE SENTIMENT COUNTS

print("\nSTEP 8: CALCULATING SENTIMENT COUNTS")

positive_count = (tweet_df["TextBlob_Sentiment"] == "Positive").sum()
negative_count = (tweet_df["TextBlob_Sentiment"] == "Negative").sum()
neutral_count = (tweet_df["TextBlob_Sentiment"] == "Neutral").sum()

print("\nSTEP 8 OUTPUT: OVERALL SENTIMENT SUMMARY")

print("Total Tweets:", len(tweet_df))
print("Positive Tweets:", positive_count)
print("Negative Tweets:", negative_count)
print("Neutral Tweets:", neutral_count)

# STEP 9: SENTIMENT COUNT TABLE

print("\nSTEP 9 OUTPUT: SENTIMENT COUNTS")

sentiment_counts = tweet_df["TextBlob_Sentiment"].value_counts()
display(sentiment_counts.to_frame(name="Number of Tweets"))

# STEP 10: SENTIMENT BAR CHART

print("\nSTEP 10: SENTIMENT BAR CHART")

plt.figure(figsize=(8, 5))

sns.countplot(
    x="TextBlob_Sentiment",
    data=tweet_df,
    order=["Positive", "Negative", "Neutral"]
)

plt.title("Sentiment Distribution of COVID-19 Tweets")
plt.xlabel("Sentiment")
plt.ylabel("Number of Tweets")
plt.show()

# STEP 11: SENTIMENT PIE CHART

print("\nSTEP 11: SENTIMENT PIE CHART")

sentiment_values = [
    positive_count,
    negative_count,
    neutral_count
]

labels = ["Positive", "Negative", "Neutral"]

plt.figure(figsize=(7, 7))

plt.pie(
    sentiment_values,
    labels=labels,
    autopct="%1.2f%%",
    explode=(0.1, 0, 0),
    shadow=True,
    startangle=90
)

plt.title("Percentage Distribution of COVID-19 Tweet Sentiments")
plt.show()

# STEP 12: WORD CLOUD FOR ONE SAMPLE TWEET

print("\nSTEP 12: WORD CLOUD FOR SAMPLE TWEET")

text = tweet_df["clean_tweets"].iloc[n]

if text.strip():
    wordcloud = WordCloud(
        width=800,
        height=400,
        max_words=50,
        background_color="white"
    ).generate(text)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Most Frequent Words in Sample Tweet")
    plt.show()
else:
    print("Sample tweet contains no words.")

# STEP 13: PREPARE TEXT FOR WORD CLOUD

print("\nSTEP 13: PREPARING TEXT FOR WORD CLOUD")

def prepCloud(topic_text, topic):
    topic = str(topic).lower()
    topic = re.sub(r"[^a-zA-Z\s]", " ", topic)
    topic_words = set(topic.split())

    stopwords = set(STOPWORDS)
    stopwords.update(topic_words)

    text_new = " ".join(
        word for word in topic_text.split()
        if word not in stopwords
    )

    return text_new

# STEP 14: WORD CLOUD FOR ALL TWEETS

print("\nSTEP 14: WORD CLOUD FOR ALL COVID-19 TWEETS")

Topic = "COVID19"

text_all = " ".join(tweet_df["clean_tweets"].tolist())
text_new_all = prepCloud(text_all, Topic)

if text_new_all.strip():
    wordcloud = WordCloud(
        stopwords=STOPWORDS,
        max_words=100,
        max_font_size=70,
        width=1200,
        height=600,
        background_color="white"
    ).generate(text_new_all)

    plt.figure(figsize=(14, 7))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Most Frequently Used Words in All COVID-19 Tweets")
    plt.show()
else:
    print("No words available for word cloud.")

# STEP 15: WORD CLOUD FOR POSITIVE TWEETS

print("\nSTEP 15: WORD CLOUD FOR POSITIVE TWEETS")

text_positive = " ".join(
    tweet_df.loc[
        tweet_df["TextBlob_Sentiment"] == "Positive",
        "clean_tweets"
    ].tolist()
)

text_new_positive = prepCloud(text_positive, Topic)

print("Number of Positive Tweets:", positive_count)

if text_new_positive.strip():
    wordcloud = WordCloud(
        stopwords=STOPWORDS,
        max_words=100,
        max_font_size=70,
        width=1200,
        height=600,
        background_color="white"
    ).generate(text_new_positive)

    plt.figure(figsize=(14, 7))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Most Frequently Used Words in Positive COVID-19 Tweets")
    plt.show()
else:
    print("No positive tweet words available.")

# STEP 16: WORD CLOUD FOR NEGATIVE TWEETS

print("\nSTEP 16: WORD CLOUD FOR NEGATIVE TWEETS")

text_negative = " ".join(
    tweet_df.loc[
        tweet_df["TextBlob_Sentiment"] == "Negative",
        "clean_tweets"
    ].tolist()
)

text_new_negative = prepCloud(text_negative, Topic)

print("Number of Negative Tweets:", negative_count)

if text_new_negative.strip():
    wordcloud = WordCloud(
        stopwords=STOPWORDS,
        max_words=100,
        max_font_size=70,
        width=1200,
        height=600,
        background_color="white"
    ).generate(text_new_negative)

    plt.figure(figsize=(14, 7))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Most Frequently Used Words in Negative COVID-19 Tweets")
    plt.show()
else:
    print("No negative tweet words available.")

# STEP 17: FINAL SUMMARY

print("\nSTEP 17 OUTPUT: FINAL ANALYSIS SUMMARY")

print("Total Tweets:", len(tweet_df))
print("Positive Tweets:", positive_count)
print("Negative Tweets:", negative_count)
print("Neutral Tweets:", neutral_count)

print("\nCOVID-19 tweet sentiment analysis completed successfully.")
```
