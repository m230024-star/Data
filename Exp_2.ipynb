```python

# EXPERIMENT 2: Apply Various Data Analysis and Visualization Techniques on EHR

# STEP 1: Import Required Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import folium
import warnings

from google.colab import files
from IPython.display import display
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error

warnings.filterwarnings('ignore')

print("All libraries imported successfully!")


# STEP 2: Upload EHR Datasets
print("\nUpload Patient Information and Medical Conditions CSV files")

uploaded = files.upload()
file_names = list(uploaded.keys())

if len(file_names) < 2:
    raise ValueError("Please upload both CSV files.")


# STEP 3: Automatically Identify Both Datasets
datasets = []

for filename in file_names:
    data = pd.read_csv(filename)
    data.columns = data.columns.str.strip().str.upper()
    datasets.append(data)

patient_index = next(
    (i for i, data in enumerate(datasets)
     if 'ID' in data.columns and
     ('BIRTHDATE' in data.columns or 'GENDER' in data.columns)),
    None
)

condition_index = next(
    (i for i, data in enumerate(datasets)
     if 'PATIENT' in data.columns and
     'DESCRIPTION' in data.columns),
    None
)

if patient_index is None or condition_index is None:
    raise ValueError("Could not identify both datasets. Please upload the correct patient and conditions CSV files.")

dataframe = datasets[patient_index]
dataframe1 = datasets[condition_index]

dataframe = dataframe.rename(columns={'ID': 'PATIENTID'})
dataframe1 = dataframe1.rename(columns={'PATIENT': 'PATIENTID'})

print("\n========== PATIENT INFORMATION ==========")
display(dataframe.head())
print("Patient Dataset Shape:", dataframe.shape)

print("\n========== MEDICAL CONDITIONS ==========")
display(dataframe1.head())
print("Conditions Dataset Shape:", dataframe1.shape)


# STEP 4: Display Dataset Data Types
print("\n========== PATIENT DATA TYPES ==========")
print(dataframe.dtypes)

print("\n========== CONDITIONS DATA TYPES ==========")
print(dataframe1.dtypes)


# STEP 5: Visualize Patient Locations on World Map
if 'LAT' in dataframe.columns and 'LON' in dataframe.columns:
    location_data = dataframe.dropna(subset=['LAT', 'LON']).copy()

    fig = px.scatter_geo(
        location_data,
        lat='LAT',
        lon='LON',
        hover_name='CITY' if 'CITY' in location_data.columns else None,
        title='Patient Locations on World Map'
    )

    fig.update_layout(title_x=0.5)
    fig.show()
else:
    print("Latitude and Longitude columns are not available.")


# STEP 6: Visualize Patient Locations on Interactive Map
if 'LAT' in dataframe.columns and 'LON' in dataframe.columns:
    location_data = dataframe.dropna(subset=['LAT', 'LON']).copy()

    fig = px.scatter_mapbox(
        location_data,
        lat='LAT',
        lon='LON',
        hover_name='ADDRESS' if 'ADDRESS' in location_data.columns else None,
        hover_data=['CITY'] if 'CITY' in location_data.columns else None,
        size='HEALTHCARE_EXPENSES' if 'HEALTHCARE_EXPENSES' in location_data.columns else None,
        zoom=3,
        height=600
    )

    fig.update_layout(mapbox_style='open-street-map')
    fig.update_layout(margin={'r': 0, 't': 40, 'l': 0, 'b': 0})
    fig.show()


# STEP 7: Data Visualization - Number of Patients in Each City
if 'CITY' in dataframe.columns:
    plt.figure(figsize=(18, 6))
    sns.countplot(
        x='CITY',
        data=dataframe,
        order=dataframe['CITY'].value_counts().index[:20]
    )
    plt.title('Number of Patients in Each City (Top 20)')
    plt.xticks(rotation=90)
    plt.xlabel('City')
    plt.ylabel('Number of Patients')
    plt.show()


# STEP 8: Data Visualization - Medical Conditions
if 'DESCRIPTION' in dataframe1.columns:
    plt.figure(figsize=(18, 6))
    sns.countplot(
        y='DESCRIPTION',
        data=dataframe1,
        order=dataframe1['DESCRIPTION'].value_counts().index[:20]
    )
    plt.title('Most Common Medical Conditions (Top 20)')
    plt.xlabel('Number of Occurrences')
    plt.ylabel('Medical Conditions')
    plt.show()


# STEP 9: Clean the Data by Removing Unwanted Columns
unwanted_columns = ['DRIVERS', 'SUFFIX', 'MAIDEN', 'PREFIX', 'PASSPORT']

dataframe = dataframe.drop(
    columns=unwanted_columns,
    errors='ignore'
)

print("\n========== AFTER REMOVING UNWANTED COLUMNS ==========")
display(dataframe.head())
print("Patient Dataset Shape:", dataframe.shape)


# STEP 10: Remove Duplicate Records
print("\nDuplicate Patient Records Before Cleaning:",
      dataframe.duplicated().sum())

print("Duplicate Condition Records Before Cleaning:",
      dataframe1.duplicated().sum())

dataframe = dataframe.drop_duplicates()
dataframe1 = dataframe1.drop_duplicates()

print("\n========== AFTER REMOVING DUPLICATES ==========")
print("Patient Records:", len(dataframe))
print("Condition Records:", len(dataframe1))


# STEP 11: Integrate Both Datasets
df = pd.merge(
    dataframe,
    dataframe1,
    on='PATIENTID',
    how='inner'
)

print("\n========== INTEGRATED HEALTHCARE DATASET ==========")
display(df.head())
print("Integrated Dataset Shape:", df.shape)


# STEP 12: Display Integrated Dataset Information
print("\n========== INTEGRATED DATA TYPES ==========")
print(df.dtypes)

print("\n========== DATASET SUMMARY ==========")
display(df.describe(include='all'))


# STEP 13: Analyze Medical Conditions
print("\n========== MEDICAL CONDITION COUNTS ==========")
display(df['DESCRIPTION'].value_counts().head(20))

print("\n========== CITY-WISE PATIENT COUNTS ==========")
if 'CITY' in df.columns:
    display(df['CITY'].value_counts().head(20))


# STEP 14: Prepare Data for K-Means Clustering
dfViz = df.dropna(subset=['CITY', 'DESCRIPTION']).copy()

dfViz['CITY'] = dfViz['CITY'].astype(str)
dfViz['DESCRIPTION'] = dfViz['DESCRIPTION'].astype(str)

print("\n========== DATA FOR CLUSTERING ==========")
display(dfViz[['CITY', 'DESCRIPTION']].head())


# STEP 15: Group Medical Conditions City-Wise
types = pd.crosstab(
    dfViz['CITY'],
    dfViz['DESCRIPTION']
)

print("\n========== CITY-WISE MEDICAL CONDITION DATA ==========")
display(types.head())

if types.shape[0] < 2 or types.shape[1] < 1:
    raise ValueError("Not enough city or medical-condition data for clustering.")


# STEP 16: Apply Elbow Method to Find Optimal Clusters
distortions = []

max_clusters = min(10, len(types))
K = range(1, max_clusters + 1)

for k in K:
    kmean = KMeans(
        n_clusters=k,
        random_state=0,
        n_init=10,
        max_iter=500
    )
    kmean.fit(types)
    distortions.append(kmean.inertia_)

plt.figure(figsize=(10, 5))
plt.plot(K, distortions, 'bx-')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Distortion')
plt.title('Elbow Method')
plt.grid()
plt.show()


# STEP 17: Apply K-Means Clustering
number_of_clusters = min(7, len(types))

kmeans = KMeans(
    n_clusters=number_of_clusters,
    random_state=0,
    n_init=10,
    max_iter=500
)

cluster_labels = kmeans.fit_predict(types)

codes = pd.DataFrame({
    'CITY': types.index,
    'cluster': cluster_labels
})

print("\n========== K-MEANS CLUSTER RESULTS ==========")
display(codes.head(20))

print("Number of Clusters:", number_of_clusters)


# STEP 18: Apply PCA for Dimensionality Reduction
scaler = StandardScaler()
scaled_types = scaler.fit_transform(types)

pca = PCA(n_components=2)
pca_types = pca.fit_transform(scaled_types)

print("\n========== PCA VARIANCE EXPLAINED ==========")
print("First Component:",
      pca.explained_variance_ratio_[0] * 100, "%")

print("Second Component:",
      pca.explained_variance_ratio_[1] * 100, "%")

print("Total Variance Explained:",
      sum(pca.explained_variance_ratio_) * 100, "%")


# STEP 19: Visualize Clusters Using PCA
plt.figure(figsize=(9, 7))

scatter = plt.scatter(
    pca_types[:, 0],
    pca_types[:, 1],
    c=cluster_labels,
    cmap='rainbow',
    s=60
)

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('K-Means Clusters Using PCA')
plt.colorbar(scatter, label='Cluster')
plt.grid()
plt.show()


# STEP 20: Analyze Average Medical Conditions in Each Cluster
cluster_data = types.copy()
cluster_data['cluster'] = cluster_labels

print("\n========== AVERAGE MEDICAL CONDITIONS BY CLUSTER ==========")
display(cluster_data.groupby('cluster').mean())


# STEP 21: Visualize Medical Conditions by Cluster
top_conditions = types.sum().sort_values(ascending=False).head(4).index

for condition in top_conditions:
    plt.figure(figsize=(8, 5))

    sns.barplot(
        x='cluster',
        y=condition,
        data=cluster_data,
        errorbar=None
    )

    plt.title('Occurrence of ' + condition + ' in Each Cluster')
    plt.xlabel('Cluster')
    plt.ylabel('Number of Occurrences')
    plt.show()


# STEP 22: Create USA Map Using Folium
latitude = 38.5
longitude = -95.665

map_USA = folium.Map(
    location=[latitude, longitude],
    zoom_start=4
)

print("\n========== USA MAP ==========")
display(map_USA)


# STEP 23: Prepare City Location Data
required_columns = ['CITY', 'LAT', 'LON', 'STATE', 'COUNTY']

if all(col in df.columns for col in required_columns):
    city = df[required_columns].copy()

    city = city.dropna(subset=['CITY', 'LAT', 'LON'])

    number_of_occurrences = (
        city.groupby('CITY')
        .agg(
            Count=('CITY', 'size'),
            LAT=('LAT', 'mean'),
            LON=('LON', 'mean'),
            STATE=('STATE', 'first'),
            COUNTY=('COUNTY', 'first')
        )
        .reset_index()
    )

    number_of_occurrences = number_of_occurrences.merge(
        codes,
        on='CITY',
        how='left'
    )

    print("\n========== CITY-WISE OCCURRENCES ==========")
    display(number_of_occurrences.head(20))


    # STEP 24: Folium Map - Number of Medical Condition Occurrences
    map_occurrences = folium.Map(
        location=[latitude, longitude],
        zoom_start=4
    )

    n_mean = number_of_occurrences['Count'].mean()

    for _, row in number_of_occurrences.iterrows():
        folium.CircleMarker(
            location=[row['LAT'], row['LON']],
            radius=max(3, row['Count'] / n_mean * 5),
            color='blue',
            fill=True,
            fill_color='blue',
            fill_opacity=0.6,
            tooltip=(
                str(row['CITY']) +
                " | Occurrences: " + str(row['Count'])
            )
        ).add_to(map_occurrences)

    print("\n========== MAP OF MEDICAL CONDITION OCCURRENCES ==========")
    display(map_occurrences)


    # STEP 25: Folium Map - City-Wise K-Means Clusters
    map_clusters = folium.Map(
        location=[latitude, longitude],
        zoom_start=4
    )

    color_list = [
        'red', 'blue', 'green', 'purple',
        'orange', 'darkred', 'cadetblue'
    ]

    for _, row in number_of_occurrences.iterrows():
        cluster_number = int(row['cluster'])
        cluster_color = color_list[cluster_number % len(color_list)]

        folium.CircleMarker(
            location=[row['LAT'], row['LON']],
            radius=6,
            color=cluster_color,
            fill=True,
            fill_color=cluster_color,
            fill_opacity=0.8,
            tooltip=(
                str(row['CITY']) +
                " | Cluster: " + str(cluster_number)
            )
        ).add_to(map_clusters)

    print("\n========== MAP OF CITY-WISE CLUSTERS ==========")
    display(map_clusters)

else:
    print("\nCity location columns are not available.")
    print("Folium city maps could not be generated.")


# STEP 26: Final Output
print("\n========== EXPERIMENT COMPLETED ==========")
print("Patient records:", len(dataframe))
print("Medical condition records:", len(dataframe1))
print("Integrated records:", len(df))
print("Cities clustered:", len(types))
print("Clusters created:", number_of_clusters)
print("\nAll Data Analysis and Visualization Techniques Applied Successfully!")
```
